```python
#Task 6:
import pandas as pd

scores = [88, 92, 79, 93, 85]
players = ['P1','P2','P3','P4','P5']

s = pd.Series(scores, index=players)

# Rank players
ranks = s.rank(ascending=False)

# Top 3 players
top3 = s.sort_values(ascending=False).head(3)

# Lowest score player
lowest = s.idxmin()

print(ranks)
print(top3)
print(lowest)
```
