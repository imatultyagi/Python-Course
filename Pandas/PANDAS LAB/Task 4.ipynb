```python
#Task 4:
import pandas as pd

marks = [95, 82, 67, 58, 73, 49, 88]
students = ['S1','S2','S3','S4','S5','S6','S7']

s = pd.Series(marks, index=students)

# Function to assign grades
def grade(x):
    if x >= 90:
        return 'A'
    elif x >= 75:
        return 'B'
    elif x >= 60:
        return 'C'
    else:
        return 'D'

# Apply grading
grades = s.apply(grade)

# Count students in each grade
grade_count = grades.value_counts()

print(grades)
print(grade_count)
```
