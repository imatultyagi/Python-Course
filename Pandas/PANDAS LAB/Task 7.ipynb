```python
#Task 7:
import pandas as pd

names = ['anuj', 'rahul', 'sneha', 'kiran', 'amit']
s = pd.Series(names)

# Convert to uppercase
upper_names = s.str.upper()

# Find names containing 'a'
contains_a = s[s.str.contains('a')]

# Replace 'anuj' with 'Anuj Kumar'
s_replaced = s.replace('anuj', 'Anuj Kumar')

print(upper_names)
print(contains_a)
print(s_replaced)
```
