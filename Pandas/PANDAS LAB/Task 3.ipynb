```python
#Task 3 :
import pandas as pd

data = [10, None, 30, None, 50, 60, None]
s = pd.Series(data)

# Count missing before filling
missing_before = s.isna().sum()

# Fill missing values with mean
mean_val = s.mean()
s_filled = s.fillna(mean_val)

# Count missing after filling
missing_after = s_filled.isna().sum()

# Drop remaining nulls (if any)
s_clean = s_filled.dropna()

print(missing_before, missing_after)
print(s_clean)
```
