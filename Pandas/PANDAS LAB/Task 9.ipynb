```python
#TAsk 9:
import pandas as pd

sales = [200, 450, 300, 150, 500, 700, 100]
days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']

s = pd.Series(sales, index=days)

# Best and worst day
best_day = s.idxmax()
worst_day = s.idxmin()

# Total and average sales
total = s.sum()
avg = s.mean()

# Apply 20% bonus where sales > 400
bonus = s.apply(lambda x: x * 1.2 if x > 400 else x)

# Sort descending
sorted_sales = s.sort_values(ascending=False)

print(best_day, worst_day)
print(total, avg)
print(bonus)
print(sorted_sales)3
```
