```python
#Task 2:
import pandas as pd

salary = [25000, 32000, 28000, 40000, 50000]
emp = ['A', 'B', 'C', 'D', 'E']

# Use float to avoid type warnings
s = pd.Series(salary, index=emp, dtype=float)

# Increase 10% where salary < 30000
s.loc[s < 30000] = s.loc[s < 30000] * 1.10
print("After increasing 10% where salary < 30000:")
print(s)
print()

# Decrease 5% where salary > 45000
s.loc[s > 45000] = s.loc[s > 45000] * 0.95
print("After decreasing 5% where salary > 45000:")
print(s)
print()

# Total salary
print("Total salary:", s.sum())
```
