```python
#Task 5:
import pandas as pd

s1 = pd.Series([10, 20, 30], index=['A', 'B', 'C'])
s2 = pd.Series([5, 15, 25], index=['B', 'C', 'D'])

# Add both Series
result = s1 + s2   # NaN appears for unmatched index

# Replace NaN with 0 and recompute
result_fixed = s1.add(s2, fill_value=0)

print(result)
print(result_fixed)
```
